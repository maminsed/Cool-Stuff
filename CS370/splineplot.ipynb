```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
points=np.array([(-5, 2),(0, 0),(1, -3),(3, 1),(4, 2)])
T = np.array([
[2, 1 , 0, 0, 0],
[1, 12, 5, 0, 0],
[0, 2 , 6, 1, 0],
[0, 0 , 1, 6, 2],
[0, 0 , 0, 0, 1],
])
r = np.array([-6/5,-231/5,-12,12,-1])
```

```python
S = np.linalg.solve(T,r)
```

```python
delta_x = np.diff(points[:,0])
y_prime = np.diff(points[:,1]) / delta_x

A = points[:-1,1]
B = S[:-1]
C = (3 * y_prime - 2 * S[:-1] - S[1:]) / delta_x
D = (S[1:] + S[:-1] - 2 * y_prime) / delta_x**2

for i in range(4):
    print(f"a_{i+1}: {A[i]:.4f} b_i: {B[i]:.4f} c_i: {C[i]:.4f} d_i: {D[i]:.4f}")
```

```python
for i in range(4):
    x_space = np.linspace(points[i,0],points[i+1,0])
    y = A[i] + B[i] * (x_space - points[i,0]) + C[i] * (x_space - points[i,0]) ** 2 + D[i] * (x_space - points[i,0]) ** 3
    plt.plot(x_space,y)
    plt.plot(points[i,0],points[i,1],marker="o")
plt.plot(points[-1,0],points[-1,1],marker="o")
plt.show()
```
